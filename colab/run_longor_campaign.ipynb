# Longor planner-bypass A/B campaign launcher

Launcher contracts:
- `RUN_SCRIPT_A=validation/run_longor_fixed_plan_campaign.py --prompt-target plate`
- `RUN_SCRIPT_B=validation/run_longor_fixed_plan_campaign.py --prompt-target plate_seen`

## 5 — Start VLA server + health

```python
%%bash
set -euo pipefail

cd /content/Longor
VLA_LOG=/tmp/longor_vlabench_vla_server.log
VLA_PYTHON=/content/Longor/.vla-venv/bin/python

if curl -fsS http://localhost:8765/health; then
  echo
  echo "VLA server already healthy"
  exit 0
fi

pkill -f "services/run_vla_server.py" || true
for _ in $(seq 1 60); do
  pgrep -f "services/run_vla_server.py" >/dev/null || break
  sleep 1
done
if pgrep -f "services/run_vla_server.py" >/dev/null; then
  echo "leftover VLA server still running; not starting another" >&2
  exit 1
fi

if ! "$VLA_PYTHON" -c "import config" 2>/dev/null; then
  echo "VLA venv missing longor install — re-run section 3" >&2
  exit 1
fi

nohup "$VLA_PYTHON" services/run_vla_server.py >"$VLA_LOG" 2>&1 &

sleep 2
if ! pgrep -f "services/run_vla_server.py" >/dev/null; then
  echo "VLA server exited immediately:" >&2
  tail -50 "$VLA_LOG" >&2
  exit 1
fi

elapsed=0
for _ in $(seq 1 180); do
  if curl -fsS http://localhost:8765/health; then
    echo
    echo "VLA server healthy"
    exit 0
  fi
  if ! pgrep -f "services/run_vla_server.py" >/dev/null; then
    echo "VLA server exited during startup:" >&2
    tail -50 "$VLA_LOG" >&2
    exit 1
  fi
  elapsed=$((elapsed + 5))
  echo "waiting for server to start... (${elapsed}s)"
  sleep 5
done

echo "VLA server not healthy after ${elapsed}s; see $VLA_LOG" >&2
tail -50 "$VLA_LOG" >&2
exit 1
```

## 6 — Run planner-bypass A/B campaign (plate vs plate_seen)

```python
%%bash
set -uo pipefail

cd /content/Longor
TASK_NAME="select_fruit"
START_IDX=0
EPISODES=10
MAX_REPLANS=2
RUN_LOG_PLATE="/tmp/longor_${TASK_NAME}_fixed_target_plate_${START_IDX}_${EPISODES}.log"
RUN_LOG_PLATE_SEEN="/tmp/longor_${TASK_NAME}_fixed_target_plate_seen_${START_IDX}_${EPISODES}.log"

set +e
/content/Longor/.venv/bin/python validation/run_longor_fixed_plan_campaign.py \
  --task "$TASK_NAME" \
  --start-idx "$START_IDX" \
  --episodes "$EPISODES" \
  --prompt-target "plate" \
  --max-replans "$MAX_REPLANS" 2>&1 | tee "$RUN_LOG_PLATE"
PLATE_EXIT=${PIPESTATUS[0]}

/content/Longor/.venv/bin/python validation/run_longor_fixed_plan_campaign.py \
  --task "$TASK_NAME" \
  --start-idx "$START_IDX" \
  --episodes "$EPISODES" \
  --prompt-target "plate_seen" \
  --max-replans "$MAX_REPLANS" 2>&1 | tee "$RUN_LOG_PLATE_SEEN"
PLATE_SEEN_EXIT=${PIPESTATUS[0]}
set -e

echo "PLATE_EXIT_CODE=$PLATE_EXIT"
echo "PLATE_SEEN_EXIT_CODE=$PLATE_SEEN_EXIT"
echo "RUN_LOG_PLATE=$RUN_LOG_PLATE"
echo "RUN_LOG_PLATE_SEEN=$RUN_LOG_PLATE_SEEN"
grep -E "^PASS task=.*prompt_target='plate'" "$RUN_LOG_PLATE" | tail -n 1 || true
grep -E "^PASS task=.*prompt_target='plate_seen'" "$RUN_LOG_PLATE_SEEN" | tail -n 1 || true
```
